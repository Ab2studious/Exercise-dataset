```python
import pandas as pd
```

```python
df = pd.read_csv("exercise_dataset.csv")
```

```python
df.head()
```

```python
df["Calories Burn"] = pd.to_numeric(df["Calories Burn"], errors="coerce")

df = df.dropna(subset=["Calories Burn", "Exercise"])

df.head()
```

```python
df["Exercise"].unique()
```

```python
def workout_percentile(exercise, calories):
    
    workouts = df[df["Exercise"] == exercise]["Calories Burn"]
    
    percentile = (workouts < calories).mean() * 100
    
    return percentile
```

```python
print("Workout Percentile Calculator")
print()

print("Exercises available:")
for exercise in df["Exercise"].unique():
    print("-", exercise)

exercise = input("\nEnter your exercise exactly as shown above: ")

calories = float(input("How many calories did you burn? "))

result = workout_percentile(exercise, calories)

print()
print(f"Your workout burned more calories than {result:.1f}% of {exercise} workouts in the dataset.")
```
