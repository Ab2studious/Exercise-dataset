```python
import pandas as pds 
```

```python
data = pds.read_csv("exercise_dataset.csv")
```

```python
exercise = input("""
Choose an exercise:

1. Jumping Jacks
2. Torso Twist
3. Treadmill
4. Pushups
5. Pullups
6. Squats
7. Russian Twists

Enter 1-7: """)

if exercise in ["1", "2", "3", "4", "5", "6", "7"]:
    calories = float(input("How many calories did you burn? "))

    percentile = (data["Calories Burn"] < calories).mean() * 100

    print("Your percentile is:", round(percentile, 2))
else:
    print("Please enter a number from 1-7.")
```
